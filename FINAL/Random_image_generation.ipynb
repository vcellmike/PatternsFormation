generate multiple images 100x100 consisting of black background and white ellipses of random size and shapes

```python
import numpy as np
from PIL import Image, ImageDraw

# Set image size
image_size = (100, 100)

# Set number of images to generate
num_images = 100

# Set ellipse properties
min_radius = 5
max_radius = 20

# Generate multiple images
for i in range(num_images):
    # Create a black image
    image = Image.new("L", image_size, color=0)
    draw = ImageDraw.Draw(image)

    # Generate random number of ellipses
    num_ellipses = np.random.randint(1, 100)

    # Draw white ellipses
    for _ in range(num_ellipses):
        # Generate random ellipse properties
        center = (np.random.randint(0, image_size[0]), np.random.randint(0, image_size[1]))
        radius = (np.random.randint(min_radius, max_radius), np.random.randint(min_radius, max_radius))
        angle = np.random.randint(0, 360)

        # Draw the ellipse
        draw.ellipse([center[0]-0.5*radius[0], center[1]-0.5*radius[1], center[0]+0.5*radius[0], center[1]+0.5*radius[1]], fill=255, outline=255)

    # Save the image with a unique name
    image.save(f"data/generated_images/image_{i+1}.png")

    # Display the image
    #image.show()
```
