```python
import pickle
import pandas as pd
import numpy as np
```

```python
# change directories to relative paths using Images1, Images2, Images3 folders

def change_dir_cols(df):  
    df_dir = df["dir"]

    dir_1 = "/content/Generated_Images_7-2-24/"
    dir_2 = "/content/Generated Images/"
    dir_3 = "/content/05-09-2024_all_images/"

    new_dir_1 = "content/Images1/"
    new_dir_2 = "content/Images2/"
    new_dir_3 = "content/Images3/"

    for i in range(len(df_dir)):
        if df_dir[i].startswith(dir_1):
            df_dir[i] = df_dir[i].replace(dir_1, new_dir_1)
        elif df_dir[i].startswith(dir_2):
            df_dir[i] = df_dir[i].replace(dir_2, new_dir_2)
        elif df_dir[i].startswith(dir_3):
            df_dir[i] = df_dir[i].replace(dir_3,new_dir_3)
        else:
            print("unrecognized directory:", df_dir[i])
    return df
```

```python
with open("data/manual_classification_cl6.pkl", 'rb') as f:
    df = pickle.load(f) # deserialize using load()

df.head()
```

```python
df = change_dir_cols(df)

df.head()
```

```python
file_name = "data/manual_classification_cl6_updated_image_dir.pkl"

with open(file_name, 'wb') as f:
        pickle.dump(df, f)
```
