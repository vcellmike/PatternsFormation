```python
# Pre-sets
path_in1 = "./content/Images1"
path_in2 = "./content/Images2"
path_in3 = "./content/Images3"

path_out_csv = "./data/feats_extracted.csv"
```

```python
from pathlib import Path
import imagej
import pandas as pd

# --- 1. Setup ---
# Folder containing PNG images
input_dir = Path(path_in1)
output_csv = Path(path_out_csv)

# Initialize ImageJ (headless mode)
ij = imagej.init(headless=True)
IJ = ij.IJ()

# --- 2. Set all measurements ---
IJ.run(
    "Set Measurements...",
    "area mean standard min centroid center perimeter bounding fit shape feret "
    "integrated median skewness kurtosis area_fraction stack limit display "
    "redirect=None decimal=3"
)

# --- 3. Measure every image ---
rows = []
for img_path in sorted(input_dir.glob("*.png")):
    imp = IJ.openImage(str(img_path))
    if imp is None:
        print(f"Skipping {img_path.name}")
        continue
    IJ.run(imp, "Measure", "")
    rt = IJ.getResultsTable()
    cols = list(rt.getHeadings())
    data = {c: rt.getValue(c, rt.size() - 1) for c in cols}
    data["filename"] = img_path.name
    rows.append(data)
    imp.close()

# --- 4. Save to CSV ---
pd.DataFrame(rows).to_csv(output_csv, index=False)
print(f"Saved {len(rows)} measurements to {output_csv}")
```
