Using Generalized Functions

```python
#a = (max-min)/2 = amplitude
#b = (max+min)/2 = midle value
#k is an arbitrary large prime
# generalize form for a parameter X = a * sin(k*seed) + b
# for double function: X = x1 * (abs(x1 - b) <= abs(x2 - b)) + x2 * (abs(x1 - b) > abs(x2 - b))


func_dic = {"wt":[0.03,0.07,0.08,0.1,0.01,0.5,-0.12], "min":[0.0083,0.01,0.025,0.0227,0.008,0.25,-0.202],"max":[0.0517,0.21,0.175,0.1773,0.012,1.5,-0.038],"a":[0.0217,0.1,0.075,0.0773,0.002,0.625,0.082],
            "b":[0.03,0.11,0.1,0.1,0.01,0.875,-0.12]} # values for Ua,Ui,Ga,Gi,Da,Di,Ba from excel sheet

params = ["Ua","Ui","Ga","Gi","Da","Di","Ba"]

df = pd.DataFrame(func_dic,index = params)

k1 = [4252800263,7276126189,5499973489,7197249239,9765037697,5184725759,1548255677]
k2 = [8519335619,1776939599,8530579613,1909621843,8153781173,8846553293,5660641043]

df["k1"] = k1
df["k2"] = k2

df
```

```python
#a = (max-min)/2 = amplitude
#b = (max+min)/2 = midle value
#k is an arbitrary large prime
# generalize form for a parameter X = a * sin(k*seed) + b
# for double function: X = x1 * (abs(x1 - b) <= abs(x2 - b)) + x2 * (abs(x1 - b) > abs(x2 - b))

# create formula that returns output close to center value for each parameter
formulas  =[]
for i in params:
  x1 = str(str(df["a"][i]) + "*sin(" + str(df["k1"][i]) + "*Seed) +" + str(df["b"][i]))
  x2 = str(str(df["a"][i]) + "*sin(" + str(df["k2"][i]) + "*Seed) +" + str(df["b"][i]))
  formula = str("(" + x1 + ")" + "*(abs(" + x1 + "-" + str(df["b"][i]) + ") <= abs(" + x2 + "-" + str(df["b"][i]) + "))+"
          + "(" + x2 + ")" + "*(abs(" + x1 + "-" + str(df["b"][i]) + ") > abs(" + x2 + "-" + str(df["b"][i]) + "))")
  formulas.append(formula)

df["formula"] = formulas

df
```
