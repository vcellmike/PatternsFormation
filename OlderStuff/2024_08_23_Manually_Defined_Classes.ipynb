```python
# Manually Classify
# Class 0: avg value = 0 (No coverage)
# Class 1: avg_value = 255 (Full Coverage)
# Class 2: num_spots_inverse = 1  and num_spots > 1 (normal spots)
# class 3: num_spots = 1 and num_spots_inverse > 1 (inverse spots)
# class 4: num_spots and inverse_spots > 1 (tiger_stripes)
# class 5: all else
class_0 = []
class_1 = []
class_2 = []
class_3 = []
class_4 = []
class_5 = []
for i in range(feats_df.shape[0]):
  feats = feats_df.iloc[i]
  if feats["Mean"]  == 0:
    class_0.append(i)
  elif feats["Mean"] == 255:
    class_1.append(i)
  elif feats["num_spots"] >= 1 and feats["num_spots_inverted"] <= 2:
    class_2.append(i)
  elif feats["num_spots"] <=2 and feats["num_spots_inverted"] >= 1:
    class_3.append(i)
  elif feats["num_spots"] > 1 and feats["num_spots_inverted"] > 1:
    class_4.append(i)
  else:
    class_5.append(i)

# class_indices = [class_0,class_1,class_2,class_3,class_4,class_5]
class_indices = [class_0,class_1,class_2,class_3,class_4]

# class_dic = {"class_0":class_0, "class_1":class_1, "class_2":class_2, "class_3":class_3, "class_4":class_4, "class_5":class_5}

class_df_list = []

for i in list(range(len(class_indices))):
  df = feats_df.iloc[class_indices[i]]
  df.reset_index(inplace = True, drop = True)
  class_df_list.append(df)
```

```python
for i in class_df_list:
  print(i.shape)
```

```python
# print out random images from clusters

num_imgs = 100

fig, ax = plt.subplots(num_imgs, len(class_df_list))

for i in range(num_imgs):
  for j in range(len(class_df_list)):
    df = class_df_list[j]
    num = np.random.randint(low = 0, high= df.shape[0], size=1, dtype=int)[0]
    img = np.array(Image.open(df["dir"][num] + df["path"][num]))
    ax[i][j].imshow(img, cmap = "YlOrRd", vmin = 0, vmax = 255)
    ax[i][j].set_title(df["path"][num] + " cluster " + str(j) + " index " + str(num), fontsize = 8)
    ax[i][j].get_xaxis().set_visible(False)
    ax[i][j].get_yaxis().set_visible(False)


fig.set_figwidth(5*len(class_df_list))
fig.set_figheight(5 * num_imgs)
plt.show()
```
