Convert CSV to JPEG

```python
# get list of CSV file paths

directory = 'C:/Users/natha/OneDrive/Desktop/Summer 2023  Image analysis/ui_scan_CSV/'

files = [] # list of the paths in the folder
filenames = [] # list of all filenames in folder

# iterate over files in directory and add them to files
for filename in os.listdir(directory):
    f = directory + filename

    # checking if it is a file
    if os.path.isfile(f):
        files.append(f)
        filenames.append(filename)



# iterate through CSV files,load them as numpy arrays and save them as jpeg
count = 0 #filename index for naming new files
for path in files:
    arr = np.loadtxt(path, delimiter=',') # load numpy array of csv file
    img = Image.new( 'L', (200,200)) # create a new image
    pixels = img.load() # create the pixel map
    # fill in image with normalized pixel values(max = 7)
    for i in range(img.size[0]):    # for every col:
        for j in range(img.size[1]):    # For every row
            pixels[i,j] = (int(arr[j,i] * 255/7)) # set the colour accordingly
    img.save('C:/Users/natha/OneDrive/Desktop/Summer 2023  Image analysis/ui_scan_img/' + filenames[count]
             + ".jpeg" ) # save image
    count += 1
```
