```python
import openpyxl
from PIL import Image
import pandas as pd
```

```python
df = pd.read_csv("/content/drive/Shareddrives/Summer_2023_VCell_Ml/Summer 2023  Image analysis/param_scan_7-6-23/CSV/Multi_Cont_2.csv", header = None)
```

```python
df.shape
print(df.shape)
round(df.shape[0]/211)
```

```python
def xl_fill(i): # fills the blank 200x200 matrix with pixel values
    x = Image.new('L', (200, 200))  # create a new image
    pixels = x.load()  # create the pixel map
    for l in range(0,200): #col
        for m in range(0,200): #row
            pix_val = round(float(df[l][m + i*211 + 10]) * 255/7)
            pixels[l, m] = pix_val if pix_val <= 255 else 255
    return x

for i in range(round(df.shape[0]/211)): # iterates through and fills each image in xl_file, saves as Tiff
    img = xl_fill(i)
    fileName = df[0][i*211+1]
    fileName = fileName[41:-1]
    fileName = fileName.replace("'","")
    fileName = fileName.replace(" ",",")
    # print(fileName)
    # img.show()
    img.save(
        '/content/drive/Shareddrives/Summer_2023_VCell_Ml/Summer 2023  Image analysis/param_scan_7-6-23/Tiffs/' + fileName
        + ".tif")  # save image
```
